```python
c = input("Enter a number: ")
if c.isdigit():
    print(f"Double of the number you entered is: {c * 2}")
else:
    print("Please enter a valid number.")
```

```python
# Integer casting is required as the default type of input is string.

c = input("Enter a number: ")
if c.isdigit():
    print(f"Double of the number you entered is: {int(c) * 2}")
else:
    print("Please enter a valid number.")
```
