```python
a=51
b=21
if b>a:
    print(f'{b} is greater than {a}')
elif b==a:
    print(f'{b} is equal to {a}')    
else:
    print(f'{b} is less than {a}')    
```

```python
i=0
while i<4:
    print(i)
    i+=1
```

```python
i=0
while i<10:
    print(2*i +3)
    i+=3
```

For 

```python
for i in range(6):
    print(i)
```

```python
for i in range(10,22,2):
    print(i)
```

```python
cars=['Tesla', 'Mazda','Ford','Toyota','Volvo']
```

```python
for car in cars:
    print(car)
```

```python
course= 'Business Context in Data Analytics' 
```

```python
all_letters=[]

for letter in course:
    all_letters.append(letter)
```

```python
print(all_letters)
```

```python
print(course.lower())
```

```python
print(course.upper())
```

```python
vowel_letters='aeiouAEIOU'
```

```python
course_vowel_letters=[letter for letter in course if letter in vowel_letters]
print(course_vowel_letters)
```

```python
course_vowel_letters=[letter for letter in course if letter not in vowel_letters]
print(course_vowel_letters)
```

```python
vowel_letters_low='aeiou'
course_vowel_letters= [letter for letter in course if letter in vowel_letters_low]
print(course_vowel_letters)
```
